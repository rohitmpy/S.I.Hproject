```python
categorical_cols = [
    "state_code",
    "dist_code",
    "court_no",
    "judge_position",
    "female_defendant",
    "female_petitioner",
    "female_adv_def",
    "female_adv_pet",
    "type_name",
    "purpose_name"
]
numeric_cols = [
    "filing_year",
    "filing_month",
    "filing_dayofweek"
]
```

```python
print(X_train[categorical_cols].dtypes)
print(X_train[numeric_cols].dtypes)
print(X_train[categorical_cols].isna().sum())
```

```python
for col in categorical_cols:
    train_values = set(X_train[col].unique())
    val_values = set(X_val[col].unique())
    test_values = set(X_test[col].unique())

    unseen_val = val_values - train_values
    unseen_test = test_values - train_values

    print(f"\n{col}")
    print("Train categories:", len(train_values))
    print("Unseen in validation:", len(unseen_val))
    print("Unseen in test:", len(unseen_test))
```

```python
# encodding for the categorical columns
category_maps = {}

for col in categorical_cols:
    unique_values = X_train[col].unique()

    # Start category IDs from 1
    mapping = {
        value: idx + 1
        for idx, value in enumerate(unique_values)
    }

    category_maps[col] = mapping
```

```python
def encode_categories(df, category_maps):
    df = df.copy()

    for col, mapping in category_maps.items():
        df[col] = df[col].map(mapping).fillna(0).astype("int32")

    return df
```

```python
X_train_encoded = encode_categories(X_train, category_maps)
X_val_encoded = encode_categories(X_val, category_maps)
X_test_encoded = encode_categories(X_test, category_maps)
```

```python
print(X_train_encoded[categorical_cols].dtypes)
print(X_train_encoded[categorical_cols].head())
```

```python
for col in categorical_cols:
    print(
        col,
        "UNK in validation:",
        (X_val_encoded[col] == 0).sum(),
        "UNK in test:",
        (X_test_encoded[col] == 0).sum()
    )
```

```python
#creating the numerical features as we need for neural network
def create_numeric_features(df):
    result = pd.DataFrame(index=df.index)

    result["filing_year"] = df["filing_year"]

    result["month_sin"] = np.sin(
        2 * np.pi * df["filing_month"] / 12
    )

    result["month_cos"] = np.cos(
        2 * np.pi * df["filing_month"] / 12
    )

    result["day_sin"] = np.sin(
        2 * np.pi * df["filing_dayofweek"] / 7
    )

    result["day_cos"] = np.cos(
        2 * np.pi * df["filing_dayofweek"] / 7
    )

    return result
```

```python
# tranforming the numerical values
X_train_numeric = create_numeric_features(X_train)
X_val_numeric = create_numeric_features(X_val)
X_test_numeric = create_numeric_features(X_test)
```

```python
# scaling the numerical values
from sklearn.preprocessing import StandardScaler

numeric_scaler = StandardScaler()

X_train_numeric = numeric_scaler.fit_transform(
    X_train_numeric
)

X_val_numeric = numeric_scaler.transform(
    X_val_numeric
)

X_test_numeric = numeric_scaler.transform(
    X_test_numeric
)
```

```python
embedding_dims = {
    "state_code": 4,
    "dist_code": 8,
    "court_no": 8,
    "judge_position": 16,
    "female_defendant": 4,
    "female_petitioner": 4,
    "female_adv_def": 4,
    "female_adv_pet": 4,
    "type_name": 32,
    "purpose_name": 32
}
```

```python
embedding_input_dims = {
    col: len(category_maps[col]) + 1
    for col in categorical_cols
}
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
categorical_inputs = {}
embeddings = []

for col in categorical_cols:

    input_layer = keras.Input(
        shape=(1,),
        name=col,
        dtype="int32"
    )

    embedding = layers.Embedding(
        input_dim=embedding_input_dims[col],
        output_dim=embedding_dims[col],
        name=f"{col}_embedding"
    )(input_layer)

    embedding = layers.Flatten()(embedding)

    categorical_inputs[col] = input_layer
    embeddings.append(embedding)
```

```python
numeric_input = keras.Input(
    shape=(5,),
    name="numeric_features",
    dtype="float32"
)
```

```python
combined = layers.Concatenate()(
    embeddings + [numeric_input]
)
```

```python
x = layers.Dense(128, activation="relu")(combined)

x = layers.Dropout(0.2)(x)

x = layers.Dense(64, activation="relu")(x)

x = layers.Dropout(0.2)(x)

x = layers.Dense(32, activation="relu")(x)

output = layers.Dense(1, name="output")(x)
```

```python
model = keras.Model(
    inputs=list(categorical_inputs.values()) + [numeric_input],
    outputs=output
)
```

```python
model.summary()
```

```python
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss=keras.losses.Huber(),
    metrics=[
        keras.metrics.MeanAbsoluteError(name="mae")
    ]
)
```

```python
train_inputs = {
    col: X_train_encoded[col].values
    for col in categorical_cols
}

train_inputs["numeric_features"] = X_train_numeric
```

```python
val_inputs = {
    col: X_val_encoded[col].values
    for col in categorical_cols
}

val_inputs["numeric_features"] = X_val_numeric
```

```python
test_inputs = {
    col: X_test_encoded[col].values
    for col in categorical_cols
}

test_inputs["numeric_features"] = X_test_numeric
```

```python
history = model.fit(
    train_inputs,
    y_train,
    validation_data=(val_inputs, y_val),
    epochs=3,
    batch_size=4096,
    callbacks=[
        keras.callbacks.EarlyStopping(
            monitor="val_loss",
            patience=2,
            restore_best_weights=True
        )
    ]
)
```

```python
pred_log = model.predict(test_inputs)

pred_days = np.expm1(pred_log.ravel())
```

```python
actual_days = test["delays_days"].values
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(actual_days, pred_days)

rmse = np.sqrt(
    mean_squared_error(actual_days, pred_days)
)

r2 = r2_score(actual_days, pred_days)

print("MAE:", mae)
print("RMSE:", rmse)
print("R2:", r2)
```

```python
print("Predicted:")
print(pd.Series(pred_days).describe())

print("\nActual:")
print(pd.Series(actual_days).describe())
```

```python
baseline_prediction = np.median(np.expm1(y_train))

baseline_pred = np.full(
    len(actual_days),
    baseline_prediction
)

print("Baseline prediction:", baseline_prediction)
print("Baseline MAE:",
      mean_absolute_error(actual_days, baseline_pred))
print("Baseline RMSE:",
      np.sqrt(mean_squared_error(actual_days, baseline_pred)))
print("Baseline R2:",
      r2_score(actual_days, baseline_pred))
```
